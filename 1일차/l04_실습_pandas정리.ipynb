## 1. 딕셔너리가 든 리스트는 바로 표가 됩니다
API 응답의 목록 데이터는 대부분 "딕셔너리가 여러 개 든 리스트" 모양으로 옵니다.
이것을 `pd.DataFrame()`에 넣으면 바로 표가 됩니다.

| 파이썬 | 표 |
|---|---|
| 리스트 전체 | 표 전체 |
| 딕셔너리 하나 | **한 행** |
| 키 | **열 이름** |
| 값 | 칸의 값 |

```python
import pandas as pd   # pd라는 짧은 이름으로 부르는 것이 관례

# 샘플 데이터(지어낸 값)
rows = [
    {"제목": "영화 가", "관객": 1200},
    {"제목": "영화 나", "관객": 3400},
    {"제목": "영화 다", "관객": 560},
    {"제목": "영화 라", "관객": 2100},
]
df = pd.DataFrame(rows)   # df = DataFrame의 줄임. 표 변수 이름으로 자주 씀

print(df.head(2))         # 앞 2행만 보기 (괄호를 비우면 5행)
print(len(df))            # 행 수
print(list(df.columns))   # 열 이름들
```

> **노트북 팁** · 셀의 마지막 줄에 `df`만 적으면 `print` 없이 예쁜 표로 보입니다.

```python
df
```

> **자주 하는 실수** · 어떤 딕셔너리에 키 하나가 빠지면 그 칸은 `NaN`(빈 값)이 됩니다. 행마다 같은 키가 있는지 확인합니다.

## 2. 열 꺼내기·붙이기, 엑셀 저장·읽기, 정렬
| 쓰는 법 | 하는 일 |
|---|---|
| `df["열"]` | 열 하나 꺼내기 |
| `df["새열"] = 리스트` | 새 열 붙이기. **리스트 길이 = 행 수** |
| `df.to_excel(파일, index=False)` | 엑셀로 저장. `index=False`는 왼쪽 번호 열(0,1,2…)을 빼라는 뜻 |
| `pd.read_excel(파일)` | 엑셀을 표로 읽기 |
| `df.sort_values("열", ascending=False)` | 큰 값부터 정렬 (`ascending=True`면 작은 값부터) |

```python
print(list(df["제목"]))                               # 열 하나 꺼내기

df["장르"] = ["코미디", "액션", "애니메이션", "드라마"]   # 4행이니 값도 4개
df.to_excel("샘플_영화.xlsx", index=False)              # 이 노트북과 같은 폴더에 저장됨

df2 = pd.read_excel("샘플_영화.xlsx")                   # 방금 저장한 파일을 다시 읽기
print(list(df2.columns))
```

### 정렬 결과는 다시 담아야 합니다
`sort_values`는 정렬한 **새 표를 돌려줄 뿐** 원래 `df`를 바꾸지 않습니다. 정렬한 표를 계속 쓰려면 `df = df.sort_values(…)`처럼 다시 담습니다.

```python
df.sort_values("관객", ascending=False)        # 정렬한 결과를 아무 데도 안 담음
print(list(df["제목"]))                        # 그대로

df = df.sort_values("관객", ascending=False)   # 다시 담음
print(list(df["제목"]))                        # 정렬됨
```

아래 셀은 **일부러 오류를 내는 예시**입니다. 행은 4개인데 값을 1개만 넣으면 길이가 맞지 않는다는 `ValueError`가 납니다.

```python
df["평점"] = [4.5]
```

## 3. 실습 과제 · 영화·공연 용어집 (60분)
`영화공연용어.xlsx`의 **용어 15개**에 위키백과 요약과 글자수를 붙여 `영화공연_용어집.xlsx`로 저장합니다.

| 단계 | 할 일 | 쓰는 것 |
|---|---|---|
| ① 엑셀 읽기 | `영화공연용어.xlsx`의 "용어" 열 | `pd.read_excel` |
| ② 요약 붙이기 | 용어마다 요약을 받아 "설명" 열에 | `get_summary()` (l03) |
| ③ 글자수 열 | 설명의 글자 수를 세어 "글자수" 열에 | `len()` |
| ④ 정렬 | 글자수가 큰 것부터 | `sort_values` |
| ⑤ 저장 | 번호 열 없이 | `to_excel(..., index=False)` |

마지막 용어 "주말문화생활추천표"는 위키백과에 없는 말이라 "설명 없음"(5글자)이 들어가고, 정렬하면 맨 아래로 갑니다.

### 3-1. get_summary 함수 준비 (l03과 같음)

```python
import time
import requests

BASE = "https://ko.wikipedia.org/api/rest_v1/page/summary/"
HEADERS = {"User-Agent": "APIClass/1.0 (me@example.com)"}

def get_summary(title):
    url = BASE + title.replace(" ", "_")
    try:
        res = requests.get(url, headers=HEADERS, timeout=5)
        res.raise_for_status()
    except requests.exceptions.RequestException:
        return "설명 없음"
    return res.json().get("extract", "설명 없음")
```

### 3-2. ① 엑셀 읽기

```python
df = pd.read_excel("영화공연용어.xlsx")
print(len(df))   # 15
df.head()
```

### 3-3. ②③ 반복하며 두 리스트 채우기
빈 리스트 두 개를 만들어 두고, 용어를 하나씩 꺼내 요약과 글자 수를 `append`로 차곡차곡 넣습니다.
15번 요청하니 10초쯤 걸립니다.

```python
summaries = []   # 요약문을 모을 빈 리스트
lengths = []     # 글자 수를 모을 빈 리스트

for term in df["용어"]:          # "용어" 열의 값을 하나씩 꺼냄
    text = get_summary(term)      # 요청 4단계를 한 번 실행
    summaries.append(text)        # 요약문 추가
    lengths.append(len(text))     # 그 요약문의 글자 수 추가
    print(term, "→", len(text), "글자")
    time.sleep(0.5)               # 다음 요청 전 0.5초 쉼
```

### 3-4. ④⑤ 열 붙이고 정렬·저장

```python
df["설명"] = summaries       # 15개 리스트 → 15행 표에 새 열
df["글자수"] = lengths
df = df.sort_values("글자수", ascending=False)   # 글자수 큰 순서로 '다시 담기'
df.to_excel("영화공연_용어집.xlsx", index=False)
print(f"{len(df)}개 용어 저장 완료")
df
```
