## 1. 요청과 응답 · 식당 비유
웹사이트에 접속하는 것도, API로 데이터를 받는 것도 모두 **요청 → 응답**입니다.

```
[내 컴퓨터]  ──요청(주소)──▶  [서버]
   브라우저 / 파이썬  ◀──응답(HTML·JSON 글자)──  데이터를 가진 컴퓨터
```

| 식당 | 웹 | 오늘의 예 |
|---|---|---|
| 손님 | **클라이언트** (요청하는 쪽) | 브라우저, 파이썬 코드 |
| 메뉴판 | **API 문서** | 어떤 항목을 요청할 수 있는지 적힌 안내 |
| 주문서 | **요청** (주소, URL) | `https://api.open-meteo.com/...` |
| 주방 | **서버** (응답하는 쪽) | Open-Meteo, 위키백과의 컴퓨터 |
| 나온 요리 | **응답** | JSON 글자 |

- 서버는 요청이 와야만 응답합니다. 먼저 말을 거는 쪽은 항상 클라이언트입니다.
- **API**는 "프로그램이 서버에 데이터를 요청하는 정해진 창구"입니다. 사람에게는 꾸민 화면(HTML)을, 프로그램에게는 데이터만(JSON) 줍니다.
- 메뉴판(API 문서)에 없는 항목은 주문할 수 없습니다. 항목 이름도 문서에 적힌 그대로 써야 합니다.

## 2. [직접 해 보기] 브라우저로 API 열기
아래 주소를 **크롬 주소창**에 붙여 넣고 엔터를 누르세요. 엔터를 누르는 순간이 곧 **요청**입니다.

https://api.open-meteo.com/v1/forecast?latitude=37.5665&longitude=126.978&current=temperature_2m

꾸민 웹 페이지가 아니라 `{"latitude":37.55, ... "temperature_2m":14.3}` 같은 **글자로 된 데이터**가 나옵니다. 이것이 JSON입니다.

## 3. URL 읽기 · 주소, 경로, 파라미터
방금 연 주소는 세 조각으로 나뉩니다.

| 조각 | 이름 | 뜻 |
|---|---|---|
| `https://api.open-meteo.com` | 주소(도메인) | 요청을 받을 서버의 이름 |
| `/v1/forecast` | 경로 | 서버 안의 어떤 기능인지 (여기서는 예보) |
| `?latitude=37.5665` | 파라미터 | `?` 뒤부터 `이름=값` 모양의 요청 내용 |
| `&longitude=126.978` | 파라미터 | `&`로 다음 파라미터를 이어 붙임 |
| `&current=temperature_2m` | 파라미터 | 받고 싶은 항목: 현재 기온 |

**규칙**: `?`는 처음에 **한 번만**, 그다음부터는 `&`로 잇습니다. 파라미터 순서는 바뀌어도 됩니다.

아래 셀은 조각을 문자열로 이어 붙여 같은 주소를 만듭니다.

````python
domain = "https://api.open-meteo.com"     # 어느 서버에
path = "/v1/forecast"                      # 서버의 어떤 기능에
query = "?latitude=37.5665&longitude=126.978&current=temperature_2m"   # 무엇을 달라고

url = domain + path + query   # 문자열은 + 로 이어 붙일 수 있음
print(url)
````

> **자주 하는 실수** · `?latitude=37.5665?longitude=126.978`처럼 `?`를 두 번 쓰면 서버가 두 번째 파라미터를 알아듣지 못합니다. 두 번째부터는 `&`입니다.

## 4. JSON은 딕셔너리·리스트 모양의 글자
서버와 내 파이썬은 서로 다른 컴퓨터라서 딕셔너리를 그대로 건넬 수 없습니다.
그래서 서버는 데이터를 **JSON 글자**로 적어 보내고, 파이썬은 그 글자를 다시 **딕셔너리**로 바꿔 씁니다.

| JSON | 파이썬 | 예 |
|---|---|---|
| `{ }` 객체 | `dict` 딕셔너리 | `{"time": "09:00"}` |
| `[ ]` 배열 | `list` 리스트 | `[12.1, 11.6]` |
| `true` / `false` | `True` / `False` | `"error": true` |
| `null` | `None` | `"description": null` |

`json.loads(글자)`가 JSON 글자를 딕셔너리로 바꿔 줍니다. (`loads` = load **s**tring)

````python
import json   # JSON을 다루는 파이썬 기본 도구 (설치 필요 없음)

# 서버가 보내 준 응답이라고 생각하세요. 아직은 '글자(str)'입니다.
text = ('{"current": {"temperature_2m": 14.3}, '
        '"hourly": {"temperature_2m": [12.1, 11.6]}}')
print(type(text))       # <class 'str'>  → 글자

data = json.loads(text)  # 글자 → 딕셔너리로 변환
print(type(data))        # <class 'dict'> → 이제 대괄호로 꺼낼 수 있음

# 딕셔너리 안의 딕셔너리는 대괄호를 이어 붙여 한 단계씩 엽니다
print(data["current"]["temperature_2m"])      # 14.3

# 리스트는 번호로 꺼냅니다. 번호는 0부터 시작
print(data["hourly"]["temperature_2m"][0])    # 12.1
````

### true · false · null은 이렇게 바뀝니다

````python
text2 = '{"ok": true, "error": false, "description": null}'
data2 = json.loads(text2)
print(data2)   # {'ok': True, 'error': False, 'description': None}
````

### 자주 하는 실수 · 키 이름의 대소문자
키 이름이 한 글자라도 다르면 `KeyError`가 납니다. 아래 셀은 **일부러 오류를 내는 예시**입니다.
응답에 적힌 키를 그대로 복사해 쓰는 습관을 들이세요.

````python
data["Current"]   # 'current'가 아니라 'Current' → KeyError
````
