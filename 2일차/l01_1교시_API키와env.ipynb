## 1. API 키란
키는 **누가 요청했는지 알려 주는 값**입니다. 식당 비유로는 주문서에 적는 **단골 회원 번호**입니다.

| 식당 | API | KOBIS 예 |
|---|---|---|
| 회원 가입 | 키 발급 | kobis.or.kr에서 신청 |
| 주문서의 회원 번호 | 요청 속 키 | `key=(내 키)` |
| 회원인지 확인 | 키 확인 | 틀리면 오류 안내 |
| 주문 횟수 세기·제한 | 요청 수 집계·한도 | 하루 요청 수 한도 |

- 키는 **요청마다 함께** 보냅니다.
- 키는 **발급한 곳에서만** 통합니다. KOBIS 키는 KOBIS에, KOPIS 키는 KOPIS에.
- 키는 **남에게 보이지 않게** 둡니다. 키를 코드에 적어 공유하면 다른 사람이 내 회원 번호로 주문하는 셈이고, 하루 한도도 그 사람이 써 버립니다.

## 2. 키 없는 API와 키 필요한 API
| API | 운영 | 키 | 키 파라미터 이름 | 응답 형식 |
|---|---|---|---|---|
| Open-Meteo 날씨 | Open-Meteo | 필요 없음 | - | JSON |
| 위키백과 요약 | 위키미디어 재단 | 필요 없음 (이름표만) | - | JSON |
| **KOBIS 박스오피스** | 영화진흥위원회 | 필요 | **`key`** | JSON |
| **KOPIS 공연 목록** | 예술경영지원센터 | 필요 | **`service`** | **XML** |

키를 넣는 파라미터 이름은 API마다 다릅니다. 이름은 API 문서에서 확인합니다.

## 4. .env 파일 만들기 (한 번만)
키는 코드가 아니라 **`.env` 파일**에 적습니다. 코드에는 키의 **이름**만 적습니다.

```
2일차/
├─ .env            ← 키를 적는 파일 (공유·업로드하지 않음)
├─ .env.example    ← .env의 견본 (이 파일을 복사해서 씀)
└─ l02_...ipynb    ← 코드
```

**만드는 법**
1. VS Code 탐색기에서 `.env.example`을 복사(Ctrl+C) → 붙여넣기(Ctrl+V)
2. 복사본 이름을 **`.env`**로 바꿈 (점으로 시작, 확장자 없음)
3. 안의 내용을 발급받은 키로 바꿔 저장
```
KOBIS_KEY=발급받은_KOBIS_키
KOPIS_KEY=발급받은_KOPIS_키
```
`=` 앞뒤에 띄어쓰기나 따옴표를 넣지 않습니다.

> **자주 하는 실수** · 파일 이름이 `env`나 `.env.txt`면 읽히지 않아 키가 `None`이 됩니다. 탐색기에서 확장자가 숨겨져 있으면 `.env.txt`가 `.env`처럼 보일 수 있으니 VS Code에서 이름을 확인하세요.

## 5. load_dotenv와 os.getenv로 키 꺼내기
| 쓰는 법 | 하는 일 |
|---|---|
| `from dotenv import load_dotenv` | 설치 이름은 `python-dotenv`, 불러올 때 이름은 `dotenv` |
| `load_dotenv()` | 같은 폴더의 `.env`를 읽어 **환경 변수**로 올림 |
| `os.getenv("KOBIS_KEY")` | 이름으로 값을 꺼냄. 값은 문자열(str) |
| `os.getenv("없는이름")` | `None`을 돌려줌 (오류가 나지 않음) |

**환경 변수**란 프로그램 바깥에 `이름=값` 모양으로 놓아 둔 값입니다.

````python
import os                        # 운영체제(환경 변수)를 다루는 기본 도구
from dotenv import load_dotenv

load_dotenv()                    # .env 파일을 읽음

kobis_key = os.getenv("KOBIS_KEY")
kopis_key = os.getenv("KOPIS_KEY")
print(type(kobis_key))           # 제대로 읽혔으면 <class 'str'>, 아니면 <class 'NoneType'>
````

## 6. 키 확인 · 요청 보내기 전에 걸러 내기
키가 `None`이면 `.env`가 없거나 이름에 오타가 있는 것입니다. 키 전체를 화면에 띄우지 않고 앞 4글자만 보여 줍니다.

````python
for name in ["KOBIS_KEY", "KOPIS_KEY"]:
    key = os.getenv(name)
    if key is None:
        print(f"❌ .env에 {name}가 없습니다")
    else:
        print(f"✅ {name}:", key[:4] + "****")   # 키 전체는 숨김
````

## 7. 키가 틀리면 어떻게 되나
키가 없거나 틀리면 **데이터 대신 오류 안내 글**이 옵니다. 상태 코드는 200(성공)으로 오기도 하니, 응답 안을 확인해야 합니다.

**KOBIS (JSON)**
```
{"faultInfo": {"message": "유효하지않은 키값입니다.", "errorCode": "320010"}}
```
**KOPIS (XML)**
```
<dbs><db><returncode>02</returncode><errmsg>SERVICE KEY IS NOT REGISTERED ERROR</errmsg>…</db></dbs>
```
그래서 l02부터는 **③ 확인하기**에서 응답 안에 `faultInfo`(KOBIS)나 `returncode`(KOPIS)가 있는지도 봅니다.
